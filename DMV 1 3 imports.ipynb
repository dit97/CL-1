```python
#combines (concatenates) them into one large DataFrame
unified_data = pd.concat([csv, ed], ignore_index=True)
```

```python
#total sales value from the SALES column of the combined dataset.
total_sales = unified_data['SALES'].sum()
print(total_sales)
```

```python
#Calculates the average (mean) sales for each order number.
category_sales = unified_data.groupby('ORDERNUMBER')['SALES'].mean()
print(category_sales)
```

```python
category_counts = unified_data['PRODUCTLINE'].value_counts()
category_counts.plot(kind='bar', color='skyblue', figsize=(8,4))
plt.title('Product Category Distribution')
plt.xlabel('Product Category')
plt.ylabel('Number of Orders')
plt.show()
```

```python
mean_sales = unified_data['SALES'].mean()

# Median
median_sales = unified_data['SALES'].median()

# Mode
mode_sales = unified_data['SALES'].mode()[0]  # mode() returns a Series, take first value

print("Mean of SALES:", mean_sales)
print("Median of SALES:", median_sales)
print("Mode of SALES:", mode_sales)
```

```python
#Each slice of the pie shows how much each order contributes to total sales.
category_sales = unified_data.groupby('PRODUCTLINE')['SALES'].sum()
category_sales.plot(kind='pie', autopct='%1.1f%%')
plt.title('Sales Share by Product Category')
plt.ylabel('')
plt.show()
```
